```python
import pandas as pd
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt

ROOT_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

DATASET_PATH = ROOT_DIR / "dataset" / "raw" / "career_navigator_synthetic.csv"
df = pd.read_csv(DATASET_PATH)

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", 200)

print("Dataset loaded successfully!")
```

```python
print("Number of records:", df.shape[0])
print("Number of columns:", df.shape[1])
print("Dataset size:", df.memory_usage(deep=True).sum() / (1024 ** 2), "MB")

print("\nColumn names:")
print(df.columns.tolist())
```

```python
display(df.head())
display(df.tail())
```

```python
print("Dataset Information:")
df.info()

print("\nData Types:")
display(df.dtypes.to_frame("Data Type"))
```

```python
missing = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df) * 100).round(2)
})

display(missing[missing["Missing Count"] > 0])
```

```python
numerical_df = df.select_dtypes(include=np.number)

display(numerical_df.describe().T)
```

```python
categorical_df = df.select_dtypes(include=["object", "category"])

for column in categorical_df.columns:
    print(f"\n{column}")
    print(df[column].value_counts())
    print("-" * 50)
```

```python
target_columns = [
    "placement_status",
    "career_readiness",
    "recommended_career_domain",
    "employability_score"
]

for column in target_columns:
    print(f"\nTarget: {column}")

    if df[column].nunique() <= 20:
        display(df[column].value_counts().to_frame("Count"))
        display(
            (df[column].value_counts(normalize=True) * 100)
            .round(2).to_frame("Percentage")
        )
    else:
        display(df[column].describe())
```

```python
REPORT_PATH = ROOT_DIR / "reports" / "eda" / "dataset_inspection.txt"
REPORT_PATH.parent.mkdir(parents=True, exist_ok=True)

with open(REPORT_PATH, "w", encoding="utf-8") as file:
    file.write("CAREER NAVIGATOR - DATASET INSPECTION REPORT\n")
    file.write("=" * 60 + "\n\n")

    file.write(f"Dataset Shape: {df.shape}\n")
    file.write(f"Memory Usage: {df.memory_usage(deep=True).sum() / (1024 ** 2):.2f} MB\n\n")

    file.write("COLUMN INFORMATION\n")
    file.write(str(df.dtypes) + "\n\n")

    file.write("MISSING VALUES\n")
    file.write(str(df.isnull().sum()) + "\n\n")

    file.write("NUMERICAL SUMMARY\n")
    file.write(str(numerical_df.describe().T) + "\n\n")

    file.write("CATEGORICAL DISTRIBUTIONS\n")
    for column in categorical_df.columns:
        file.write(f"\n{column}\n")
        file.write(str(df[column].value_counts()) + "\n")

print(f"Inspection report saved: {REPORT_PATH}")
```
